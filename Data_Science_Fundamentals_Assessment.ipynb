## 3. Statistical Foundations

This section covers mean, median, mode, variance, standard deviation, percentiles, probability, and correlation.


```python
import statistics
import numpy as np
import pandas as pd

scores = [55, 60, 65, 70, 70, 75, 80, 85, 90]

summary = pd.Series({
    "mean": np.mean(scores),
    "median": np.median(scores),
    "mode": statistics.mode(scores),
    "population_variance": np.var(scores),
    "population_std": np.std(scores),
    "25th_percentile": np.percentile(scores, 25),
    "50th_percentile": np.percentile(scores, 50),
    "75th_percentile": np.percentile(scores, 75),
})
summary
```

```python
# Sample variance and standard deviation
print("Sample variance:", np.var(scores, ddof=1))
print("Sample standard deviation:", np.std(scores, ddof=1))
```

### Probability

For equally likely outcomes:

**P(A) = favorable outcomes / total outcomes**

Example: the probability of drawing an Ace from a standard 52-card deck is 4/52 = 1/13.


```python
probability_ace = 4 / 52
print("P(Ace) =", probability_ace)
```

### Correlation

Pearson correlation ranges from -1 to +1. It measures the direction and strength of a linear association. Correlation does not by itself establish causation.


```python
study_hours = [1,2,3,4,5,6,7,8]
exam_scores = [52,55,61,65,69,74,78,84]
r = np.corrcoef(study_hours, exam_scores)[0,1]
print("Pearson correlation:", round(r, 3))
```

## 4. Data Types and Structures

- **Numerical:** quantities such as age or score
- **Categorical:** labels such as department
- **Ordinal:** ordered categories such as Low/Medium/High
- **Datetime:** dates and timestamps
- **Text:** natural-language content


```python
data = pd.DataFrame({
    "age": [19,20,21,22,20],
    "department": ["AI","Data","AI","Data","AI"],
    "satisfaction": ["High","Medium","High","Low","Medium"],
    "date_joined": pd.to_datetime(["2026-01-10","2026-02-12","2026-03-15","2026-04-20","2026-05-18"]),
    "feedback": ["Good course","Useful","Very useful","Needs improvement","Good"]
})
data["department"] = data["department"].astype("category")
data["satisfaction"] = pd.Categorical(
    data["satisfaction"],
    categories=["Low","Medium","High"],
    ordered=True
)
print(data)
print("\nData types:\n", data.dtypes)
```

## 5. Data Cleaning and Manipulation

```python
raw = pd.DataFrame({
    "student_id": [101,102,103,104,104],
    "study_hours": [2.5,4.0,np.nan,5.5,5.5],
    "attendance": [80,92,75,np.nan,np.nan],
    "score": [62,78,70,88,88],
    "department": ["AI","AI","Data","Data","Data"]
})
print(raw)
print("\nMissing values:\n", raw.isna().sum())
print("\nDuplicate rows:", raw.duplicated().sum())
```

```python
clean = raw.drop_duplicates().copy()
clean["study_hours"] = clean["study_hours"].fillna(clean["study_hours"].median())
clean["attendance"] = clean["attendance"].fillna(clean["attendance"].median())

high_scores = clean[clean["score"] >= 75]
sorted_data = clean.sort_values("score", ascending=False)
group_summary = clean.groupby("department", observed=True).agg(
    average_score=("score","mean"),
    average_attendance=("attendance","mean"),
    students=("student_id","count")
).reset_index()

print("Cleaned data:\n", clean)
print("\nHigh scores:\n", high_scores)
print("\nSorted:\n", sorted_data)
print("\nGroup summary:\n", group_summary)
```

## 6. Exploratory Data Analysis

A reproducible synthetic dataset is used so the project can run without downloading external data. The fixed random seed makes the demonstration reproducible.


```python
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 100

eda = pd.DataFrame({
    "study_hours": np.clip(rng.normal(5,1.7,n),1,10),
    "attendance": np.clip(rng.normal(82,8,n),50,100),
    "assignments_completed": rng.integers(3,11,n)
})
eda["score"] = np.clip(
    30 + 5.5*eda["study_hours"] + 0.25*eda["attendance"] + rng.normal(0,8,n),
    0, 100
)
eda.head()
```

```python
plt.figure(figsize=(7,4))
plt.hist(eda["score"], bins=10, edgecolor="black")
plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.scatter(eda["study_hours"], eda["score"], alpha=0.7)
plt.title("Study Hours vs Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.show()
```

```python
print(eda.describe().round(2))
print("\nCorrelation matrix:\n", eda.corr(numeric_only=True).round(2))
```
